#### 2. What's the distribution of your most important column? 

Here, I decided to show the distribution of the ratings across every review. In addition to the distribution, my code also returned the number of reviews in the dataframe, the mean of the ratings, its standard deviation, minimum, maximum, and percentages. 

```python
# Distribution of star ratings (1–5): counts and share of all reviews
rating_counts = df["rating"].value_counts().sort_index()
rating_pct = df["rating"].value_counts(normalize=True).sort_index().mul(100).round(1)

dist = pd.DataFrame({"count": rating_counts, "percent": rating_pct})
print(dist)

df["rating"].describe().round(3)
```

#### 3. Filter to a meaningful subset. What's in it? 
I filtered the dataset to show only reviews that reviewed Dovetail and rated it at 3 or lower. With this filtered dataframe, I can see that most of these ratings are at 3, and very few of them are rated 1 or 2. This makes me wonder if Dovetail actually has a greater amount of reviews rating it higher.

```python
# Dovetail reviews with rating 3 or lower (3, 2, or 1)
dovetail_low = df[(df["app"] == "Dovetail") & (df["rating"] <= 3)]

print(len(dovetail_low), "rows")
dovetail_low
```

---
#### 4. Group by category and find the average numeric column.
Here, I decided to group the dataframe by app and selected Miro, where I took the average of all its ratings.

```python
# Average rating for rows where app is Miro
miro_reviews = df[df["app"] == "Miro"].copy()
miro_avg_rating = miro_reviews["rating"].mean()

# Sort Miro reviews by rating (low to high), then by date
miro_sorted = miro_reviews.sort_values(["rating", "date"], ascending=[True, True]).copy()

print("Miro average rating:", round(miro_avg_rating, 2))
print("Miro rows:", len(miro_sorted))

# Show sorted Miro table
miro_sorted
```

---
#### 5. Where are the missing values? Are any columns incomplete? 
Building on the exercise above, the code cell below flags any missing values and counts the number of incomplete columns for each review that includes Miro as its app of choice. Based on this analysis, we can see that the columns that most commonly had missing values were "app version" and "device type", which may indicate that users either typically don't keep track of their app versions, or have a difficult time finding that information. 

```python
# 5) Missing values in the Miro subset: flag rows and count incomplete columns
miro_missing_check = miro_sorted.copy()
miro_missing_check["has_missing_values"] = miro_missing_check.isnull().any(axis=1)
miro_missing_check["incomplete_column_count"] = miro_missing_check.isnull().sum(axis=1)

print("Rows with at least one missing value:", int(miro_missing_check["has_missing_values"].sum()))

# Missing count per column/category (descending)
missing_by_column = miro_missing_check.isnull().sum().sort_values(ascending=False)
print("\nMissing values by column (descending):")
print(missing_by_column)

miro_missing_check
```
